```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
epoch1 = np.loadtxt('W1049A_wRes_E1.txt', delimiter=' ')
epoch2 = np.loadtxt('WISE1049A_Median.txt', delimiter=' ')
```

```python
epoch1[:,0]
```

```python
# Plot the data
plt.figure(figsize=(10, 5))
plt.plot(epoch1[:, 0], epoch1[:, 1], label='Epoch 1 Median', color='blue')
plt.plot(epoch2[:, 0], epoch2[:, 1], label ='Epoch 2 Median', color='orange')
plt.xlabel('Wavelength (microns)')
plt.ylabel('Flux (w/m^2/um)')
plt.title('WISE 1049A')
plt.legend()
plt.xlim(0.5, 10)
plt.grid()
```
